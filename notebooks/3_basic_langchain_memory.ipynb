```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_anthropic import ChatAnthropic

llm = ChatAnthropic(model="claude-opus-5")
```

### Manage history

For this purpose:

    * We will create a history[]

    - We will append the user, and Ai messages (i.e. question and answer)
    
    - In this way the memroy will be saved in history[]

```python
history = []

while True:
    user_input = input("User: ")

    # condition to break the loop
    if user_input.lower() in ["exit", "quit", "bye"]:
        print("Exiting the chat. Goodbye!")
        break

    history.append({"role": "user", "content": user_input})
    print("User: ", user_input)

    # we will not invoke "User_input" here,
    # rather we will invoke "history" coz we have to save the history
    res = llm.invoke(history)
    history.append({"role": "ai", "content": res.content})
    print("AI: ", res.content, "\n")
```
