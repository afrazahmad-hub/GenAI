```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_community.utilities import GoogleSerperAPIWrapper
from langchain_anthropic import ChatAnthropic
from langchain.agents import create_agent
from langgraph.checkpoint.memory import MemorySaver # for memory
```

```python
# create an object of "Google Serper"
search = GoogleSerperAPIWrapper()

# llm model
model = ChatAnthropic(model="claude-opus-5")
```

```python
Search_answer = search.run("First neame of Obama?")
print(Search_answer)
```

```python
agent = create_agent(
    model=model,
    tools=[search.run],
    system_prompt="You are and agent to anser the questions by searching on Google.",
    checkpointer=MemorySaver() # memory
)
```

```python
question = "Full name of Imran khan Pakistan?"

res = agent.invoke({
    "messages": [
        {
            "role": "user",
            "content": question
        }
    ]},
    # we can change the "thread Value" according to specific user e.g for x,y,z to preserve there specific history
    {"configurable": {"thread_id": "1"}}, 
)

print(res["messages"][-1].content)
```

```python
for r in res["messages"]:
    print(r)
    print("\n")
    
```
