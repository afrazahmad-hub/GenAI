# Dynamic Prompts

```python
# to create reusable blueprints for AI conversations.
from langchain_core.prompts import ChatPromptTemplate
# the method extract only the string output from the result (i.e. res.content)
from langchain_core.output_parsers import StrOutputParser

output = StrOutputParser()
prompts =  ChatPromptTemplate.from_messages([
    {"role": "system", "content": "you will translate message in {language}"},
    {"role": "user", "content": "{query}"}
])

# Solely for us......just to print responce in uppercase(PANGA)
def to_upperCase(result: str):
    return result.upper()
```

#### Method 1 to print answer

```python
# formating the message using `formate_message` class,
# but problem is we have always change the language and query variables
final_prompt = prompts.format_messages(language = "French",
                        query = "I love eating")

res = model.invoke(final_prompt)
print(f"French Translation: {res.content}")
```

#### Method 2 to print answer

### Runables

```python
# Both are runables
prp = prompts.invoke({"language": "French", "query": "I'm going to sleep."}).messages
res = model.invoke(prp)

# print(prp)
print(f"French Translation: {res.content}")
```

#### Method 3 to print answer

### Chains

```python
# We can summaries the all code of above 2 cells in following 1 line.
# It works as
# model then print it....same sequence as above cell
# Prompt data -> model -> string output  
chains = prompts | model | output | to_upperCase
res = chains.invoke({"language": "French", "query": "I just finished the work."})
print(res)
```
