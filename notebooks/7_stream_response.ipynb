```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_anthropic import ChatAnthropic

llm = ChatAnthropic(model="claude-sonnet-4-5-20250929", streaming=True)
```

```python
res = llm.stream("Who is the president of France?")
```

```python
# we must have to create "stream" object next to "for" loop
# Coz, each time it has to create an object

for chunk in res:
    print(chunk.content, end="")
```
