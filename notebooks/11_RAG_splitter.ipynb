## Documents and Text Split in RAG

- We use `langchain-text-splitters` to split the docs/text

### 1. Text Split

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

# we can change the aruguments "chunk_size" and "chunk_overlap" according to the length of data
splitter = RecursiveCharacterTextSplitter(chunk_size=60, chunk_overlap=20)

texts = "Public speaking can be daunting for students. They often struggle to start, " \
"engage the audience, and be memorable. It's a fear of forgetting words or losing the audience's interest. " \
"This leads to anxiety and self-doubt."

text = splitter.split_text(texts)

print(len(text))
text
```

### 2. Split Document

- to split docs we first need `loader` to load the document.

```python
# loading docs

# mports
from langchain_community.document_loaders import TextLoader
from langchain_community.document_loaders import PyPDFLoader

# Create objects
text_loader = TextLoader("../data/speech.txt")
pdf_loader = PyPDFLoader("../data/BESAVO_Businessplan.pdf")

# uploading files
text_file = text_loader.load()
pdf_file = pdf_loader.load()
```

```python
# Splitting text file 
splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=40)
text_split = splitter.split_documents(text_file)
```

```python
print(len(text_split))

# created 8 Lists of text document, according to the given chunk size & overlap.
text_split
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size=100, chunk_overlap=60)

pdf_split = splitter.split_documents(pdf_file)

print(len(pdf_split))
# created 121 Lists of documents, according to the chunk size we gave.
pdf_split
```
