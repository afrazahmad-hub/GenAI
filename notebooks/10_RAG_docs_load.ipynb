## Uploading the files for RAG

- A single Langchain method i.e. `langchain_community.document_loaders` can upload various types of files (i.e. text, pdf, csv, etc.)

### Loading the Text file

```python
from langchain_community.document_loaders import TextLoader
loader = TextLoader("../data/speech.txt") # haven't yet load at this stage
text = loader.load() # loaded
one_doc = text[0]
print(one_doc.page_content)
```

### Loading the PDF file

```python
from langchain_community.document_loaders import PyPDFLoader
loader = PyPDFLoader("../data/BESAVO_Businessplan.pdf")
doc = loader.load()

print("Total Pages: ", len(doc))
print(doc[0])
```

### Load CSV file

```python
from langchain_community.document_loaders import CSVLoader
loader = CSVLoader("../data/bank-additional-full.csv")
data = loader.load()
print(len(data))
# get first 5 data points
print(data[:5])
```

### Load Website 

```python
# For this we have to install the beautifulsoup4 packege
# which is website hhtp data crawler
from langchain_community.document_loaders import WebBaseLoader
loader = WebBaseLoader(web_path="https://www.cashxchain.com/en/pricing")
data = loader.load()
print(data[0].page_content)
```

## Langchain Built-in packeges

- We will test Wikipedia by installing it

```python
from langchain_community.document_loaders import WikipediaLoader
# if we dont specify the load_max_docs it will load all possibkle pages and consume more tokens
loader = WikipediaLoader(query="France?", load_max_docs=2) 
wikipd = loader.load()
len(wikipd)
print(wikipd[0].page_content)
```
