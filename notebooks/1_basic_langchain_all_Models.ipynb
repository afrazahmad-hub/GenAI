### Langchain Anthropic

```python
from langchain_anthropic import ChatAnthropic

model = ChatAnthropic(model="claude-opus-5", temperature=0.9)

res = model.invoke("What is the oldest city of France?")
print(res.content)
```

### Langchain Google Generative AI

```python
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(model="gemini-3-flash-preview")

res = llm.invoke("In short tell me what is GenAI?")
print(res.text)
```

# Static Prompt

```python
from langchain_anthropic import ChatAnthropic
llm = ChatAnthropic(model="claude-opus-5", temperature=0.5)

prompt = [
    ("system", "You are a senior python developer"),
    ("user", "How to sort arrays?")
]

res = llm.invoke(prompt)
print(res.content)
```
