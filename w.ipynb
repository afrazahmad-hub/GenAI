## 1. What is an LLM?

**LLM = Large Language Model**

An LLM is a machine-learning model trained on a very large amount of text data to understand and generate human language.

### Meaning of "Large Language Model"

* **Large** → trained using a huge amount of data.
* **Language** → focuses on human language/text.
* **Model** → a mathematical/computational model capable of learning patterns from data.

Training data can include:

* Books
* Articles
* Websites
* Programming-related text
* Other large collections of text available for training

The model learns patterns from this data and uses those learned patterns to generate responses to user input.

---

## 2. What is GPT?

**GPT = Generative Pre-trained Transformer**

The three parts describe the model:

### Generative

The model can generate new content based on the input it receives.

### Pre-trained

The model is trained beforehand on large amounts of data before being used for tasks such as answering questions or generating text.

### Transformer

A neural-network architecture designed to process language effectively and capture relationships between different parts of an input.

---

## 3. Transformer Architecture

Transformers were introduced in the 2017 research paper:

> **"Attention Is All You Need"**

The paper introduced the **attention mechanism**, which became a fundamental component of Transformer-based language models.

### Why is attention important?

Words can have different meanings depending on their surrounding context.

For example:

> "The bank was two steps away from the river."

The word **bank** could refer to:

* A financial institution
* The side of a river

The attention mechanism allows the model to consider relationships between words in the input and determine which surrounding words are important for interpreting a particular word.

### Key idea

**Attention → relationships between tokens → better understanding of context → improved text generation**

---

## 4. How LLMs Generate Text

A simplified view of text generation is **next-token prediction**.

Suppose the input is:

> "I went to the"

The model evaluates possible continuations, such as:

* beach
* bench
* chair
* floor

It assigns probabilities to possible next tokens and selects a token according to the model's generation process.

The selected token is then added to the existing input, and the model predicts the next token again.

### Simplified generation loop

```text
Input
  ↓
Predict next token
  ↓
Add predicted token to input
  ↓
Predict next token
  ↓
Repeat
  ↓
Final generated response
```

For example:

```text
"I went to the"
        ↓
"I went to the beach"
        ↓
"I went to the beach and"
        ↓
"I went to the beach and sat"
        ↓
...
```

The process continues until the model determines that generation should stop.

### Important

LLMs do not simply retrieve a complete answer from their training data. Their generation process involves predicting subsequent tokens based on the context and learned patterns.

# 5. What are Tokens?

Computers and language models do not directly process human-readable sentences in the same way humans do.

Before text can be processed by an LLM, it is converted into **tokens**.

A token is a unit of text used by the model. Depending on the tokenizer, a token can represent:

* A whole word
* Part of a word
* Punctuation
* Spaces or combinations involving spaces
* Other pieces of text

For example:

```text
Who is the PM of India?
```

may be split into multiple tokens such as:

```text
Who
 is
 the
 PM
 of
 India
?
```

The exact tokenization depends on the tokenizer/model.

---

# 6. Token IDs

Tokens themselves are still text representations. LLMs therefore map tokens to numerical identifiers.

Each token is assigned a unique **token ID** according to the model's vocabulary.

Example:

```text
Tokens:
["Who", " is", " the", " PM", " of", " India", "?"]

Token IDs:
[... numerical IDs ...]
```

The LLM ultimately processes these numerical representations.

### Important relationship

```text
Human text
    ↓
Tokens
    ↓
Token IDs
    ↓
LLM
    ↓
Output token IDs
    ↓
Tokens
    ↓
Human-readable text
```

---

# 7. Tokenization

**Tokenization** is the process of converting text into tokens.

A **tokenizer** performs this conversion.

### Simplified process

```text
"Who is the PM of India?"
            ↓
        Tokenization
            ↓
["Who", " is", " the", " PM", " of", " India", "?"]
            ↓
       Token IDs
            ↓
[...numbers...]
```

The model processes the token IDs rather than the original raw text.

---

# 8. Model-Specific Tokenizers

Different LLMs can use different tokenizers and vocabularies.

Therefore, the same text can produce different:

* Tokens
* Token IDs
* Number of tokens

depending on the model/tokenizer being used.

For example, changing between different GPT model/tokenizer versions can result in different token IDs for the same input.

### Key point

> **Always consider the tokenizer associated with the model you are using.**

This is particularly important when calculating:

* Context usage
* API costs
* Maximum input/output size
* Token limits

# 10. Tokenization with Python

The lesson demonstrates tokenization using the Python **`tiktoken`** library.

### Install/import

```python
import tiktoken
```

A tokenizer is created for a particular model:

```python
tokenizer = tiktoken.encoding_for_model("MODEL_NAME")
```

Then text can be converted into token IDs:

```python
token_ids = tokenizer.encode(text)
print(token_ids)
```

### Decode token IDs back to text

```python
result = tokenizer.decode(token_ids)
print(result)
```

So the two fundamental operations are:

```text
encode()  → text → token IDs
decode()  → token IDs → text
```

### Example workflow

```python
text = "The PM of India is Modi."

tokenizer = tiktoken.encoding_for_model("MODEL_NAME")

token_ids = tokenizer.encode(text)
print(token_ids)

result = tokenizer.decode(token_ids)
print(result)
```

The decoded result should reconstruct the original text represented by those token IDs.

# 11. Context Window

A **context window** defines how much token-based information an LLM can handle within a request/conversation context.

It concerns the amount of:

* Input tokens the model can receive
* Output tokens the model can generate

A simplified representation:

```text
Context Window
┌───────────────────────────────┐
│ Input tokens + Output tokens  │
└───────────────────────────────┘
```

If a conversation becomes very long, the available context can eventually become a limitation.

### Practical implication

When an application's conversation or input becomes too large for the model's context window, the application must manage the context, for example by reducing or summarizing information.

---

# 12. Maximum Token Limits

LLM APIs typically impose limits on the number of tokens that can be processed or generated.

When building an LLM application, it is important to understand:

* Maximum input tokens
* Maximum output tokens
* Total context window
* Model-specific limits

These limits affect both application design and cost.

---

# 13. Token-Based Pricing

Many commercial LLM APIs charge based on token usage.

There can be separate costs for:

* **Input tokens** → tokens sent to the model
* **Output tokens** → tokens generated by the model

Therefore:

```text
Higher token usage
       ↓
Higher API cost
```

The exact price depends on the provider and model.

### Why token optimization matters

For an LLM application, reducing unnecessary tokens can:

* Reduce API costs
* Reduce unnecessary context
* Help stay within model limits
* Improve application efficiency

This is why concepts such as token limits and token optimization become important when developing production LLM applications.
