# Introduction to AI Agents

## 1. What is an LLM?

A **Large Language Model (LLM)** primarily works by processing a user's prompt and generating a text-based response.

For example:

```text
User → Question → LLM → Text Response
```

However, an LLM by itself generally does not perform external actions such as:

* Accessing a database
* Calling external APIs in real time
* Searching Google or the web
* Retrieving current weather information
* Booking a flight
* Creating or updating records in an external system

This creates a limitation: an LLM can generate instructions or explain how to perform a task, but it cannot necessarily perform that task on its own.

### Example

If a user asks:

> "Book me a flight from Bhopal to Mumbai."

A basic LLM may explain how to book a flight, but it cannot actually complete the booking without access to an appropriate external system.

---

# 2. Tools

**Tools** provide LLMs with the ability to interact with external systems and perform actions.

In a Python-based AI application, a tool can be represented by a **Python function**.

For example, different tools could provide:

* Database access
* Weather information
* Google/web search
* API calls
* Task creation
* Other application-specific operations

Conceptually:

```text
Python Function → Tool → External System
```

The function performs the actual operation and returns the result.

### Example

A weather tool might:

```text
LLM
 ↓
Weather Tool
 ↓
Weather API
 ↓
Current Temperature
 ↓
LLM
```

The LLM can then use the returned information to formulate a response for the user.

---

# 3. Connecting LLMs with Tools

When tools are connected to an LLM, the LLM can determine whether it needs to use a tool based on the user's request.

For example:

```text
User:
"What is the current temperature in Bhopal?"

        ↓

      LLM
        ↓
Does the LLM already have the required information?
        ↓
      No
        ↓
Select Weather Tool
        ↓
Weather API
        ↓
Tool Response
        ↓
      LLM
        ↓
Human-readable response
        ↓
      User
```

The LLM therefore acts as a decision-making layer between the user and the available tools.

---

# 4. Tool Selection

An LLM can have access to multiple tools.

For example:

```text
Tool 1 → Database Access
Tool 2 → Weather API
Tool 3 → Web Search
```

When the user asks a question, the LLM determines which tool is appropriate.

For example:

* "Find the current temperature" → Weather Tool
* "Find information on the web" → Search Tool
* "Create a task in the system" → Database/Task Tool

The LLM generates the required tool call and arguments, the tool performs the operation, and its result is returned to the LLM.

---

# 5. The Think → Act → Observe Pattern

An agent can be understood through three main stages:

### 1. Think

The LLM analyzes the user's request and determines what needs to be done.

It decides whether:

* It can answer directly, or
* It needs to perform an external action.

### 2. Act

If an external action is required, the LLM selects an appropriate tool and calls it with the necessary arguments.

### 3. Observe

The LLM receives and analyzes the tool's response.

It determines whether the result satisfies the user's request.

If the result is insufficient, the agent can repeat the cycle:

```text
Think
  ↓
Act
  ↓
Observe
  ↓
Think
  ↓
Act
  ↓
Observe
  ...
```

This continues until an appropriate final response can be produced.

---

# 6. ReAct Pattern

The **ReAct (Reasoning + Action)** pattern describes the process of combining reasoning with actions.

The basic cycle is:

```text
Reasoning → Action → Observation
```

The model reasons about what should happen, performs an action through a tool, observes the result, and can then reason again based on that result.

ReAct is therefore a commonly used pattern for building tool-using AI systems and agents.

> **ReAct ≠ React.js**

ReAct in AI refers to **Reasoning + Action**, not the JavaScript/TypeScript frontend library React.

---

# 7. What is an AI Agent?

An **AI agent** can be understood as a system that combines:

1. **Prompt/User Input**
2. **LLM**
3. **Tools**

Conceptually:

```text
            User
              ↓
          User Prompt
              ↓
             LLM
          ↙       ↘
      Reasoning   Tools
                    ↓
              External Systems
                    ↓
               Tool Result
                    ↓
                   LLM
                    ↓
              Final Response
```

The important difference is that an agent is not limited to simply generating text. It can use tools to interact with external systems and perform tasks.

---

# 8. LLM vs AI Agent

| LLM                                               | AI Agent                                                |
| ------------------------------------------------- | ------------------------------------------------------- |
| Primarily generates text                          | Can perform actions                                     |
| Works mainly with the information available to it | Can obtain external information through tools           |
| Does not inherently access external systems       | Can interact with APIs, databases, search systems, etc. |
| Usually responds directly to a prompt             | Can reason, use tools, observe results, and continue    |
| Text generation is the main capability            | Task execution is an important capability               |

An agent therefore extends the capabilities of an LLM by connecting it with tools and an execution workflow.

---

# 9. Building AI Agents with LangChain

There are several ways to build AI agents. Different LLM providers and AI frameworks provide agent-building capabilities.

In this series, **LangChain** is used as the framework for developing agents with Python.

A basic agent requires:

```text
User Prompt
     +
    LLM
     +
   Tools
     ↓
   Agent
```

The user prompt provides the task, the LLM provides the reasoning and decision-making capability, and the tools allow the agent to interact with external systems.

---

# 10. Key Takeaways

* An **LLM** primarily generates text based on user input.
* An LLM alone cannot inherently perform arbitrary external actions.
* **Tools** extend an LLM's capabilities by allowing it to interact with external systems.
* In Python, tools can be implemented using **functions**.
* An LLM can select an appropriate tool based on the user's request.
* Tool results are returned to the LLM for further processing.
* The **Think → Act → Observe** cycle is a fundamental pattern for tool-using agents.
* **ReAct** stands for **Reasoning + Action** and describes this iterative approach.
* An **AI agent** combines an LLM, user input, and tools to perform tasks.
* **LangChain** can be used to build and orchestrate AI agents in Python.
* Understanding how tools are created is an important step toward building agents.
