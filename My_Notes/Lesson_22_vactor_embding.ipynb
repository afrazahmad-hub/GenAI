# Vector Embeddings and Vector Databases

## 1. Vector Embeddings

**Vector embeddings** are numerical representations of text or other data. They convert information into a list of numbers that captures aspects of its semantic meaning and relationships.

Unlike simple token IDs, embeddings help represent semantic similarities between different pieces of text.

### Why Are Embeddings Important?
- Convert text into numerical representations that AI systems can process.
- Help identify relationships between words, sentences, and documents.
- Enable semantic similarity searches, even when the wording is different.

## 2. Vector Space and Similarity Search

Each embedding represents a data point in a multidimensional vector space. Texts with similar meanings can have similar vector representations.

**Cosine similarity** is a common method for comparing two vectors by measuring the angle between them. It helps determine how similar their representations are.

Other comparison methods include Euclidean distance and dot product.

## 3. Embedding Models

Embedding models transform raw text, such as sentences or paragraphs, into fixed-length numerical vectors.

Different embedding models can produce vectors with different dimensions. The model chosen affects the representation of text and may influence retrieval quality.

Examples mentioned in the tutorial include OpenAI embeddings, Google Generative AI embeddings, and Ollama-supported embeddings.

## 4. Vector Databases

A **vector database** stores embeddings and supports similarity-based searches to retrieve semantically relevant information.

Traditional SQL or NoSQL databases are commonly used for exact-value queries, such as searching by an ID or status. Vector databases are designed to find the nearest or most semantically similar vectors.

Examples mentioned in the tutorial include ChromaDB, Astra DB, Amazon OpenSearch, FAISS, and MongoDB's vector search capabilities.

## 5. Vector Database Workflow

The tutorial describes two main phases:

### Indexing Phase
1. Prepare the source documents.
2. Convert the text into embeddings using an embedding model.
3. Store the embeddings in a vector database.
4. Persist the data when future reuse is required.

### Query Phase
1. Receive a user's query.
2. Search the vector database for similar records.
3. Return the most relevant results.
4. Optionally limit the number of results using **Top-K retrieval**, which specifies how many matching results to return.

## 6. Vector Databases in RAG

In a Retrieval-Augmented Generation (RAG) system, vector databases help retrieve relevant document chunks based on a user's question. The retrieved information can then be provided to an LLM as context for generating an answer.

This allows the model to answer questions using relevant external information rather than relying only on its existing training knowledge.

## Key Takeaways

- **Embeddings:** Numerical representations that capture semantic meaning.
- **Embedding models:** Convert text into vectors of a fixed dimension.
- **Cosine similarity:** A common way to compare vector similarity.
- **Vector databases:** Store embeddings and support semantic retrieval.
- **Top-K retrieval:** Selects a specified number of the most similar results.
- **RAG:** Uses retrieved information to provide context for LLM-generated answers.
