# 01. What is an LLM?

**LLM = Large Language Model**

An LLM is a machine-learning model trained on a very large amount of text data to understand and generate human language.

### What does "Large Language Model" mean?

- **Large** → trained on a large amount of data.
- **Language** → works with human language/text.
- **Model** → a computational model that learns patterns from data.

### Typical Training Data

- Books
- Articles
- Websites
- Programming-related text
- Other large text datasets

# 02. What is GPT?

**GPT = Generative Pre-trained Transformer**

### Generative
Can generate new content based on the provided input.

### Pre-trained
The model is trained on large amounts of data before being used by users.

### Transformer
A neural-network architecture designed to process language and model relationships between different parts of the input.

# 03. Transformer Architecture

Transformers were introduced in the 2017 research paper:

**"Attention Is All You Need"**

A key concept introduced in the architecture is the **attention mechanism**.

The attention mechanism allows the model to consider relationships between different words/tokens in the input.

### Why does this matter?

The meaning of a word can depend on its surrounding context.

Example:

> The bank was near the river.

The word **bank** can have different meanings. The surrounding context helps determine which meaning is relevant.

# 05. How LLMs Generate Text

A simplified view of LLM generation is **next-token prediction**.

Example:

```text
"I went to the"
        ↓
Predict next token
        ↓
"I went to the beach"
        ↓
Predict next token
        ↓
"I went to the beach and"
        ↓
Continue...

At each step, the model predicts possible next tokens based on the existing context.

The selected token is added to the sequence, and the process repeats until generation stops.



# 06. What are Tokens?

A **token** is a unit of text processed by an LLM.

A token can represent:

- A complete word
- Part of a word
- Punctuation
- Other pieces of text

Example:

```text
Who is the PM of India?

# 07. Token IDs

```text

 ↓
Tokens
 ↓

For example:

["Who", " is", " the", " PM", " of", " India", "?"]
             
             ↓
        [numerical IDs]

The LLM processes these numerical representations.


# 08. Tokenization

**Tokenization** is the process of converting text into tokens.

```text
"Who is the PM of India?"
            ↓
        Tokenizer
            ↓
        Tokens
            ↓
        Token IDs

The tokenizer determines how the input text is divided.

# 09. Tokenizer & Vocabulary

Different LLMs can use different tokenizers and vocabularies.

Therefore, the same text may produce different:

- Tokens
- Token IDs
- Token counts

depending on the model/tokenizer.

This matters when working with:

- Context windows
- API costs
- Token limits
- Input/output size

# 10. Tokenization with Python

The `tiktoken` library can be used to encode text into token IDs and decode token IDs back into text.

Two important operations:

```text
encode() → text → token IDs

decode() → token IDs → text

````python
import tiktoken

tokenizer = tiktoken.encoding_for_model("gpt-4")

# print the tokens in the following string
text = "What is the capital of France?"

tokens = tokenizer.encode(text)
print(f"Tokens: {tokens}")

# from the tokens to text
Tokens = [3923, 374, 279, 6864, 315, 9822, 30]
result = tokenizer.decode(Tokens)
print(f"Decoded text: {result}")
````

# 11. Context Window

The **context window** represents how much token-based information a model can handle within its context.

It includes the tokens associated with the model's input and the generated output, subject to the model's limits.

```text
┌───────────────────────────────┐
│        Context Window         │
│                               │
│ Input tokens + Output tokens  │
│                               │
└───────────────────────────────┘


### Why it matters

Long conversations or large documents can eventually exceed the available context.

Applications therefore need strategies for managing large amounts of context.

# 12. Maximum Token Limits

LLM APIs impose model-specific token limits.

When developing an LLM application, consider:

- Maximum input tokens
- Maximum output tokens
- Total context window
- Model-specific restrictions

These limits affect both application design and cost.

# 13. Token-Based Pricing

LLM APIs commonly measure usage in tokens.

Costs can depend on:

- **Input tokens** → tokens sent to the model
- **Output tokens** → tokens generated by the model

```text
More tokens
    ↓
More usage
    ↓
Potentially higher API cost
