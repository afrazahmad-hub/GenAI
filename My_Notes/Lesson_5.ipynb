# Lesson 5 — LangChain: LLM Integration and Prompts

## 1. LangChain and LLM Integration

**LangChain** is a framework that allows applications to work with different Large Language Models (LLMs) through a common framework.

Instead of designing an application around only one LLM provider, LangChain provides integrations for multiple providers.

Examples discussed in the lesson include:

* OpenAI
* Anthropic
* Google Gemini
* Azure
* Amazon Bedrock

### Why this is useful

Different LLM providers expose their models through different APIs and interfaces. LangChain provides a more consistent way of interacting with these models.

This means an application can change the underlying LLM while keeping much of the application structure unchanged.

---

# 2. Models in LangChain

The **model** is a central component of LangChain because it is responsible for connecting the application to an LLM.

LangChain provides integrations for different model providers.

The general workflow is:

```text
Application
     ↓
LangChain
     ↓
Selected LLM Provider
     ↓
LLM Model
     ↓
Response
```

The specific provider and model can change, while the overall interaction pattern remains similar.

---

# 3. Different LLM Providers

LangChain supports models from multiple providers through provider-specific integrations.

The lesson demonstrates this concept using:

* **Google Gemini**
* **OpenAI**

The same general interaction pattern can also be applied to other supported providers such as Anthropic.

### Important concept

The application primarily needs to change the **model/provider integration**, rather than completely redesigning how it sends inputs and receives outputs.

This abstraction is one of the useful aspects of using LangChain.

---

# 4. Standardized Model Interaction

One important advantage of LangChain is that different LLM integrations follow a similar interaction pattern.

For example, after an appropriate model object has been created, the application can generally:

1. Provide an input.
2. Invoke the model.
3. Receive a response.
4. Extract the generated content.

The underlying provider may be different, but the application-level interaction can remain largely consistent.

### Conceptual comparison

```text
Application
    │
    ├── OpenAI Model
    │
    ├── Google Gemini Model
    │
    └── Anthropic Model
```

This makes it easier to experiment with or switch between different LLM providers.

These capabilities allow LLMs to be used for a wide variety of applications.

# 6. Prompts

A **prompt** is the input provided to an LLM together with instructions that help define how the model should process that input.

A prompt is therefore more than simply a question.

It can contain:

* Instructions
* Context
* User input
* Information about the expected role or behavior of the model

### Basic idea

```text
Prompt
  ↓
Instructions + Context + User Input
  ↓
LLM
  ↓
Response
```

Prompt design becomes increasingly important when building LLM applications.

---

# 7. Types of Prompts

The lesson introduces two broad types of prompts:

1. **Static prompts**
2. **Dynamic prompts**

## Static Prompts

A static prompt has a fixed structure and fixed content.

The instructions and input do not dynamically change based on variables or changing application data.

## Dynamic Prompts

Dynamic prompts can be constructed or modified according to changing information.

For example, an application could dynamically insert user information, retrieved data, or other context into a prompt.

The lesson focuses on **static prompts** and indicates that dynamic prompts will be covered later.

---

# 8. Message Roles in Prompts

Prompts can contain different types of messages.

The lesson focuses on three roles:

* **System**
* **AI**
* **User**

Each role provides different information to the model.

---

# 9. System Prompt

A **system prompt** provides instructions about how the LLM should behave or perform a task.

It establishes the model's intended role, behavior, or operating instructions.

### Example concept

A system instruction could tell the model:

> Act as a Python developer.

The purpose is not to ask the model a specific question. Instead, it establishes the role that should guide subsequent responses.

### Why use a system prompt?

A system prompt can provide consistent behavioral instructions across user requests.

For example:

```text
System:
"You are a Python developer."

User:
"How do I sort an array?"
```

The model can use the system-level instruction to interpret the request from the perspective of a Python developer.

If the system instruction changes:

```text
System:
"You are a Java developer."
```

the same user request can be answered from a Java-development perspective.

### Key idea

**System prompt → establishes the model's role and behavior.**

---

# 10. User Prompt

The **user prompt** represents the actual request or question provided by the user.

For example:

```text
User:
"How do I sort an array?"
```

The user message communicates what the user wants the model to accomplish.

The model interprets this request together with any system instructions and other available context.

---

# 11. AI Message

An **AI message** represents a response generated by the model.

It can also be used as part of previous conversational context.

For example, in a multi-turn conversation:

```text
System → Defines behavior
User   → Asks a question
AI     → Generates an answer
User   → Asks a follow-up
```

The previous AI response can become part of the context used to process subsequent user messages.

---

# 12. Relationship Between System, User, and AI Messages

These message roles serve different purposes:

| Message Role | Purpose                                 |
| ------------ | --------------------------------------- |
| **System**   | Defines instructions, role, or behavior |
| **User**     | Provides the user's request or task     |
| **AI**       | Represents model-generated responses    |

Conceptually:

```text
System
  ↓
Defines how the model should behave

User
  ↓
Provides the task

AI
  ↓
Generates the response
```

Together, these messages provide structured context to the LLM.

---

# 13. Static Prompt Structure

A static prompt can contain fixed system instructions and a fixed user request.

For example:

```text
System:
"You are a Python developer."

User:
"How do I sort an array?"
```

The important point is that the prompt structure and its content are predetermined.

This differs from a dynamic prompt, where parts of the prompt can be generated or inserted dynamically.

---

# 14. Why System Prompts Matter

System prompts allow developers to establish a consistent role for an LLM.

For example, an application could define the model as:

* A Python developer
* A Java developer
* A frontend developer
* A data engineer
* A GenAI engineer

The user does not necessarily need to repeat this instruction in every question.

The system-level instruction establishes the context that guides subsequent responses.

### Example

```text
System:
"You are a Java developer."

User:
"How do I sort an array?"
```

The model can interpret the request in the context of Java development.

If the system instruction instead establishes a Python role, the response can be generated in the context of Python.

---

# 15. LangChain's Abstraction for Multiple Models

The lesson demonstrates that different LLM providers can be connected using similar LangChain concepts.

For example:

```text
LangChain Application
        │
        ├── Google Gemini
        │
        ├── OpenAI
        │
        └── Anthropic
```

The provider-specific model integration changes, but the overall application interaction remains similar.

This abstraction reduces the need to learn a completely different application-level interaction pattern for every LLM provider.

---

# 16. Static vs Dynamic Prompting

### Static Prompt

```text
Fixed instructions
       +
Fixed input
       ↓
     LLM
```

### Dynamic Prompt

```text
Instructions
     +
Changing application data
     +
User input
     ↓
Prompt generated dynamically
     ↓
LLM
```

Static prompts are useful when the instructions and input structure are known in advance.

Dynamic prompts become useful when an application needs to construct prompts based on changing information.

---

# 17. Key Relationships

The major concepts from the lesson can be connected as follows:

```text
LangChain
    ↓
Model Integration
    ↓
Different LLM Providers
    ↓
Prompt
    ↓
System + User + AI Messages
    ↓
LLM Processing
    ↓
Generated Response
```

LangChain provides an abstraction layer that allows applications to interact with different LLM providers using similar application-level concepts.

# Key Takeaways

* **LangChain** provides a framework for integrating different LLM providers into applications.
* LangChain supports integrations with providers such as **OpenAI, Google Gemini, Anthropic, Azure, and Amazon Bedrock**.
* The **model** is a central component responsible for interacting with an LLM.
* Different LLM providers can be accessed through provider-specific integrations while maintaining a similar application-level interaction pattern.
* **Prompts** provide instructions, context, and user input to an LLM.
* Prompts can be **static** or **dynamic**.
* A **system prompt** defines the model's intended role, behavior, or instructions.
* A **user prompt** contains the user's actual request.
* An **AI message** represents a response generated by the model and can form part of conversational context.
* System instructions can influence how the model responds to subsequent user requests.
* Static prompts use predetermined instructions and input, while dynamic prompts can incorporate changing information.
* LangChain's model abstraction makes it easier to work with or switch between different LLM providers.
