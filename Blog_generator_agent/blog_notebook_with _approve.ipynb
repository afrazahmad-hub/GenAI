```python
graph = build_blog_graph()
graph
```

```python
# Configuration
config = {"configurable": {"thread_id":"1"}}

# For Research node
state = graph.invoke(
    {
        "topic":"AML in cryptocurrencies",
        "audience":"Complience officer and begginers."
    },
    config=config
    )
```

```python
# Research stage
snap = graph.get_state(config) # get state of particular user "config"
interupt_payload = snap.interrupts[0].value

# checking the stage
print("Stage: ", interupt_payload.get("stage"))
print("\n ----------- Research Outline for User Review --------- \n")
print("Stage: ", interupt_payload.get("research"))
```

```python
# To move next stage --- approve research
# "config" will help to get the current state of particular ID,
# and then start from where it was stopped.
state = graph.invoke(
    Command(resume={"action":"approve", "feedback":""}),
    config=config
)
```

```python
state
```

```python
# Draft stage
snap = graph.get_state(config) # get state of particular user "config"
interupt_payload = snap.interrupts[0].value
# print(interupt_payload)

# checking the stage
print("Stage: ", interupt_payload.get("stage"))
print("\n ----------- Draft for User Review --------- \n")
print(interupt_payload.get("draft"))
```

### Approve for final Draft (i.e. Editor)

```python
state = graph.invoke(
    Command(resume={"action":"approve", "feedback":""}),
    config=config
)

print("\n ----- Topic  ---- \n")
print(state["topic"])
print("\n ----- Audience ---- \n")
print(state["audience"])
print("\n -----Final Blog ---- \n")
print(state["final_blog"])
```
